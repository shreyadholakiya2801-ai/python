```python
def safe_divide(a, b):
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        return "Cannot divide by zero"


# Function calls
print(safe_divide(10, 2))
print(safe_divide(10, 0))

```

```python
try:
    reviews = int(input("Enter number of reviews: "))
    total_stars = int(input("Enter total stars: "))

    if reviews <= 0 or total_stars < 0:
        print("Please enter valid positive values.")      
    else:
        average_rating = total_stars / reviews
        print("Average Rating:", average_rating)

except ValueError:
    print("Error: Please enter numeric values only.")     #abc
```

```python
def get_playlist_duration(songs):
    total = 0

    for i in songs:
        if i < 0:
            raise InvalidDurationError("Negative duration not allowed")

        total = total + i

    return total / 60


songs = [120, 180, 240]

try:
    print(get_playlist_duration(songs), "minutes")

except InvalidDurationError as e:
    print(e)
```
