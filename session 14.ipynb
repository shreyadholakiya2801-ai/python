```python
import csv

# Create a sample IPL CSV file

with open("ipl_matches.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["Match", "Team1", "Team2", "Winner"])
    writer.writerow(["Match 1", "CSK", "MI", "CSK"])
    writer.writerow(["Match 2", "RCB", "KKR", "KKR"])
    writer.writerow(["Match 3", "RR", "SRH", "RR"])
    writer.writerow(["Match 4", "GT", "DC", "GT"])

print("CSV file created successfully!")
```
