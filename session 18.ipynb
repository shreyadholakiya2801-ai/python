```python
import re

text = """
Contact: 9876543210
Price: 1500
Phone: 8765432109
WhatsApp: 7654321098
Random: 1234567890
Another: 9123456789
"""

pattern = r'\b[789]\d{9}\b'

numbers = re.findall(pattern, text)

print("Valid Indian Phone Numbers:", numbers)
```

task 2

```python
import re

def check_date(text):
    pattern = r'\b\d{2}/\d{2}/\d{4}\b'

    if re.search(pattern, text):
        return True
    else:
        return False


print(check_date("My birthday is 28/01/2027"))
print(check_date("Today is 2026-11-2"))
```

```python
import re

text = """
Customer review:
Contact: shreya@gmail.com
Email: bipin123@yahoo.com
Support: rishi@zomato.com
Invalid email: yash@gmail
Another email: ayush@gmail.com
"""

pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'

emails = re.findall(pattern, text)

print("Extracted Email Addresses:", emails)
```

task 5

```python
import re

def check_order_id(text):
    pattern = r'\bOD\d{18}\b'

    result = re.search(pattern, text)

    if result:
        return result.group()
    else:
        return "Invalid Order ID"


orders = [
    "Your order ID is OD123456789012345000",
    "Order number: OD987654321098765432",
    "Order ID: AB123456789012345000"
]

for order in orders:
    print(check_order_id(order))
```

```python
# 1. re.findall()
# findall() return all matches as a list
# \d = digit (0-9)
# + = one o0r more dogits
```

```python
import re

text = " ram second 80, shyam scord 95, mohan scord 88 ."
numbers = re.findall(r"\w+",text)
# /d = digit (0-9)
# + = one ie more digit
print(numbers)
```

```python
import re

text = " ram second 80, shyam scord 95, mohan scord 88 ."
numbers = re.findall(r"\d+",text)
# /d = digit (0-9)
# + = one ie more digit
print(numbers)
```

```python
text = " phyton is esay"
words = re.findall(r"\w+" , text)
print(words)
```

4 . common regex patterns

```python
# 1. numbers
# pattern - \d+

text = " age: 18  salary: 50000"
print(re.findall(r"\d+" , text))
```

```python
# 2. Email pattern
# A commonly used pattern

import re

text = """
abc@gmail.com
xyz@yahoo.com
hello123@hotmail.com
"""

email = re.findall(
    r"[a-zA-Z0-9._%+-]+@[a-zA-Z0-9 -]+\.[a-zA-Z]{0,3}",
    text
)

print(email)
```

```python
# 3. phone number
# indian mobile numbers(10 digits)
# pattern - \d{10}

import re

text = " call me at 6768746298 or 4546379202"
phones = re.findall(r"\d{10}" , text)
print(phones)
```
