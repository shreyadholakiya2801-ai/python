```python
def calculate_final_price(price, discount_rate):
    return price * (1 - discount_rate)


# Test the function
final_price = calculate_final_price(1200, 0.15)
print("Final Price: Rs.", final_price)
```

```python
def apply_discount(price , rate=0.10):
    discount = price * rate
    final_price = price - discount
    return final_price

result = apply_discount(2000)
print(result)
```

```python
def calculate_cashback(amount, cashback_rate = 0.05):
    cashback = amount * cashback_rate
    return cashback

zomato_cashback = calculate_cashback(500)
print("zomato_cashback:", zomato_cashback)

flipcart_cashback = calculate_cashback(2000, 0.07)
print("flipcart_cashback:", flipcart_cashback)
```
