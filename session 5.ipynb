```python
def remove_last_item(order_list):
    return order_list.pop()

# Sample Zomato order list
order_list = ["Pizza", "Burger", "Pasta", "Fries"]

removed_item = remove_last_item(order_list)

print("Removed item:", removed_item)
print("Updated order list:", order_list)
```

```python
insta_filters = ("Clarendon", "Gingham", "Juno", "Lark")

print("Original filters:", insta_filters)

try:
    insta_filters[1] = "Valencia"
except TypeError as e:
    print("Error:", e)

# Tuples are immutable, so their elements cannot be changed
# after creation. Direct assignment raises a TypeError.
```
