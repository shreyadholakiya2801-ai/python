```python
class BikeListing:

    def __init__(self, reserve_price):
        self.__reserve_price = reserve_price

    def get_reserve_price(self):
        return self.__reserve_price

    def set_reserve_price(self, value):
        if value <= 0:
            raise ValueError("Price must be greater than 0")
        self.__reserve_price = value


# Create object
bike = BikeListing(50000)

# Print price using getter
print("Reserve Price:", bike.get_reserve_price())

# Update price using setter
bike.set_reserve_price(60000)
print("Updated Price:", bike.get_reserve_price())

# print(bike.__reserve_price)  # Direct access fails
```

```python
class BikeListing:

    def __init__(self, reserve_price):
        self.reserve_price = reserve_price

    @property
    def reserve_price(self):
        return self.__reserve_price

    @reserve_price.setter
    def reserve_price(self, value):
        if value < 5000 or value > 200000:
            raise ValueError("Price must be between 5000 and 200000")
        self.__reserve_price = value


# Test 4 values
listing = BikeListing(30000)

values = [30000, 50000, 3000, 250000]

for value in values:
    try:
        listing.reserve_price = value
        print(value, "ACCEPTED")
    except ValueError as e:
        print(value, e)
```

```python
from abc import ABC, abstractmethod

class Listing(ABC):

    @abstractmethod
    def final_price(self):
        pass


# listing = Listing()  # TypeError


class BikeListing(Listing):

    def __init__(self, price):
        self.price = price

    def final_price(self):
        return self.price + (self.price * 2 / 100)


# Create object
bike = BikeListing(50000)

print("Original Price:", bike.price)
print("Final Price:", bike.final_price())
```

```python
class BikeListing:

    def __init__(self, reserve_price):
        self.reserve_price = reserve_price

    @property
    def reserve_price(self):
        return self.__reserve_price

    @reserve_price.setter
    def reserve_price(self, value):
        if value < 5000 or value > 200000:
            raise ValueError("Invalid Price")
        self.__reserve_price = value

    def final_price(self):
        return self.reserve_price + (self.reserve_price * 2 / 100)


# Test 6 prices
values = [0, -100, 500, 5000, 30000, 250000]

for i in values:
    try:
        listing = BikeListing(i)
        print(i, "ACCEPTED")
        print("Final Price:", listing.final_price())

    except ValueError:
        print(i, "REJECTED")
```
