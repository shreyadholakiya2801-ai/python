```python
def sum_playlist_durations(durations):
    if len(durations) == 0:
        return 0
    return durations[0] + sum_playlist_durations(durations[1:])

durations = [180, 200, 240, 150]

print("Total duration:", sum_playlist_durations(durations), "seconds")
```

```python
def count_likes(posts):
    total = 0

    for post in posts:
        total += post['likes']

        if 'replies' in post:
            total += count_likes(post['replies'])

    return total
```

```python

posts = [
    {
        'likes': 10,
        'replies': [
            {
                'likes': 5,
                'replies': [
                    {'likes': 3}
                ]
            },
            {'likes': 2}
        ]
    },
    {'likes': 8}
]

print(count_likes(posts))
```

```python
# Global variable
app_status = "Online"

def whatsapp_status():
    # Local variable
    user_status = "Offline"

    print("Inside function:")
    print("User Status:", user_status)
    print("App Status:", app_status)


# Before function call
print("Before function call:")
print("App Status:", app_status)

# Calling function
whatsapp_status()

# After function call
print("After function call:")
print("App Status:", app_status)

# Trying to access local variable outside function
# print(user_status)  # This will cause an error
```
