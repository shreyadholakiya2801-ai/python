```python
# example find the sum using an itertor

numbers = [7,9,5,6,2,1,4,5,6]

it = iter(numbers)
total = 0

for i in it:
    total += i 
print(total)    
```

```python
student = ["shreya","neha", "pinal"," jinal", " gita"]

for i in range(len(student)):
    print(i,student[i])


for i,j in enumerate(student):
    print(i,j) 



# 0 shreya
# 1 neha
# 2 pinal
```

zip()

```python

student = ["shreya","neha", "pinal"," jinal", " gita"]
marks = [78,56,89,12,45]
city = ["ahmd","surat","delhi","brd","amreli"]


for i,j,k in zip (student,marks,city):
    print(i,j,k) 

list(zip(student , marks , city))    
```
