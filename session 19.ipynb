```python

class Bike:
    dealer_name = "Shree Bike Showroom"

    def __init__(self, model_name, year, km_driven, price_inr):
        self.model_name = model_name
        self.year = year
        self.km_driven = km_driven
        self.price_inr = price_inr

    def apply_discount(self, percent):
        self.price_inr = self.price_inr - (self.price_inr * percent / 100)
        return round(self.price_inr)


bike1 = Bike("Honda", 2020, 25000, 55000)

print("Dealer Name:", Bike.dealer_name)

print("Before Discount:", bike1.price_inr)

bike1.apply_discount(10)

print("After Discount:", bike1.price_inr)
```

```python

class Bike:
    def __init__(self, model_name, year, km_driven, price_inr):
        self.model_name = model_name
        self.year = year
        self.km_driven = km_driven
        self.price_inr = price_inr

    def description(self):
        return self.model_name + " Price: Rs. " + str(self.price_inr)

    def is_budget(self):
        return self.price_inr < 40000


# Create 8 bikes
bikes = [
    Bike("Honda", 2020, 25000, 55000),
    Bike("TVS", 2022, 12000, 35000),
    Bike("Bajaj", 2018, 40000, 45000),
    Bike("Yamaha", 2019, 30000, 60000),
    Bike("Hero", 2017, 50000, 30000),
    Bike("Suzuki", 2021, 15000, 70000),
    Bike("Royal Enfield", 2016, 45000, 80000),
    Bike("KTM", 2023, 10000, 75000)
]

total = 0
budget_count = 0
oldest = bikes[0]

for bike in bikes:
    total = total + bike.price_inr

    if bike.is_budget():
        budget_count = budget_count + 1

    if bike.year < oldest.year:
        oldest = bike

average = total / len(bikes)

print("Average Price:", average)
print("Oldest Bike:", oldest.description())
print("Budget Bikes:", budget_count)
```
