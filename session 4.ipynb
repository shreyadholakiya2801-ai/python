```python
# lower() = badha leeter ne lowercase kare che..
# replace('_' ,' ') dash (-) ne space ma badle che.
text = 'Flipkart-sale2024'

result = text.lower().replace('_',' ')

print(result)
```

task 2

```python
# strip() = sruvat ane and ma rehli extra spaces dhur kare che.
# upeer() = badha letters ne CAPITAL banve che.
# replace('-',':') = dash ne colon (:) ma badle che.

product = ' onePlus Nord-CE 3'
result = product.strip().upper().replace('-',':')

print(result)
```

task 3

```python
# def = function banva mate vapry che.
# split_product_code(product_code) = function nu name ane parameter che.
# split('-') = dash jya hoy tya string ne alag parts ma vheche che.
# return = result pachu ape che.
# ahi result ake list che.

def split_product_code(product_code):
    return product_code.split('-')

result = split_product_code('ZOMATO-FOOD-2024')

print(result)
```
