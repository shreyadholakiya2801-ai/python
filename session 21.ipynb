task 1

```python
class Vehicle:
    def __init__(self, model_name, reg_year, price_inr):
        self.model_name = model_name
        self.reg_year = reg_year
        self.price_inr = price_inr


class Bike(Vehicle):
    def __init__(self, model_name, reg_year, price_inr, engine_cc):
        super().__init__(model_name, reg_year, price_inr)
        self.engine_cc = engine_cc


class Scooter(Vehicle):
    def __init__(self, model_name, reg_year, price_inr, boot_litres):
        super().__init__(model_name, reg_year, price_inr)
        self.boot_litres = boot_litres


# Create objects
b1 = Bike("Honda Shine", 2022, 75000, 125)
s1 = Scooter("Honda Activa", 2023, 85000, 18)

# Print Bike details
print("Bike Details:")
print(b1.model_name, b1.reg_year, b1.price_inr, b1.engine_cc)

# Print Scooter details
print("Scooter Details:")
print(s1.model_name, s1.reg_year, s1.price_inr, s1.boot_litres)
```

task 2

```python
class Vehicle:
    def __init__(self, model_name, reg_year, price_inr):
        self.model_name = model_name
        self.reg_year = reg_year
        self.price_inr = price_inr

    def resale_estimate(self):
        age = 2026 - self.reg_year
        return self.price_inr * (1 - 0.08) ** age


class Bike(Vehicle):
    def resale_estimate(self):
        age = 2026 - self.reg_year
        return self.price_inr * (1 - 0.10) ** age


class Scooter(Vehicle):
    def resale_estimate(self):
        age = 2026 - self.reg_year
        return self.price_inr * (1 - 0.07) ** age


# Create objects
b1 = Bike("Honda Shine", 2022, 75000)
s1 = Scooter("Honda Activa", 2023, 85000)

# Print resale estimates
print("Bike Resale Price:", round(b1.resale_estimate()))
print("Scooter Resale Price:", round(s1.resale_estimate()))
```

task 3

```python
# Create 6 objects
vehicles = [
    Bike("Honda Shine", 2022, 75000),
    Scooter("Honda Activa", 2023, 85000),
    Bike("Pulsar", 2020, 90000),
    Scooter("TVS Jupiter", 2021, 80000),
    Bike("Apache", 2022, 95000),
    Scooter("Suzuki Access", 2024, 90000)
]

total = 0

# Loop through all vehicles
for v in vehicles:
    estimate = v.resale_estimate()
    print(type(v).__name__, v.model_name, round(estimate))
    total = total + estimate

print("Total Resale Estimate:", round(total))
```

task 4

```python
import csv

# Sort in descending order
sorted_vehicles = sorted(
    vehicles,
    key=lambda v: v.resale_estimate(),
    reverse=True
)

# Write CSV file
with open("two_wheeler_estimates.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["type", "model", "age_years", "resale_estimate_inr"])

    for v in sorted_vehicles:
        writer.writerow([
            type(v).__name__,
            v.model_name,
            2026 - v.reg_year,
            round(v.resale_estimate())
        ])

# Read CSV file
with open("two_wheeler_estimates.csv", "r") as file:
    reader = csv.reader(file)

    rows = list(reader)

    print("CSV Data:")
    for row in rows:
        print(row)

    print("Total Data Rows:", len(rows) - 1)
```
